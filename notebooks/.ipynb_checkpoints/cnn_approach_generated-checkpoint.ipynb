
# Projet Stations 5G - Approche CNN (Convolutional Neural Network)

Ce notebook implémente l'approche basée sur un réseau de neurones convolutifs (CNN) pour la classification des stations 5G.

**Objectifs :**
1. Charger les données prétraitées depuis `data/processed/`.
2. Analyser brièvement les données sous l'angle de l'approche CNN (reshape).
3. Prétraiter les données : mise à l'échelle et remodelage (reshape) pour le format CNN.
4. Définir et compiler l'architecture CNN en utilisant les paramètres de `src/config.py`.
5. Gérer le déséquilibre des classes (pondération).
6. Entraîner le modèle CNN, sauvegarder le meilleur modèle dans `models_trained/`.
7. Évaluer les performances du modèle.


```

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
import sys
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks

# Scikit-learn imports
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, balanced_accuracy_score
from sklearn.utils import class_weight

# --- Configuration des Chemins et Modules ---
# Ajouter le dossier src au PYTHONPATH pour importer les modules perso
# Le notebook est dans 'notebooks/', donc src est '../src/'
SRC_DIR = Path('../src').resolve()
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

# Importer les modules personnalisés depuis src/
try:
    import config
    import utils
    import data_loader
    print("Modules config, utils, data_loader importés depuis src.")
except ModuleNotFoundError as e:
    print(f"ERREUR: Impossible d'importer les modules depuis {SRC_DIR}.")
    print("Vérifiez la structure du projet et la présence des fichiers __init__.py si nécessaire.")
    print(e)
    # Définir des placeholders si l'import échoue pour permettre l'exécution partielle
    class config: SEED = 42; IMG_ROWS=54; IMG_COLS=64; CNN_FILTERS = [32]; CNN_KERNEL_SIZE=(3,3); CNN_POOL_SIZE=(2,2); CNN_DENSE_UNITS=[64]; CNN_DROPOUT_RATE=0.5; CNN_ACTIVATION='relu'; CNN_OUTPUT_ACTIVATION='softmax'; CNN_OPTIMIZER='adam'; CNN_LOSS='sparse_categorical_crossentropy' # Minimal config
    class utils:
        @staticmethod
        def set_seed(seed): print(f"Seed set to {seed}")
        @staticmethod
        def plot_history(hist): print("Plotting history (dummy)")
        @staticmethod
        def plot_confusion_matrix(y_true, y_pred, classes): print("Plotting CM (dummy)")
    class data_loader:
        @staticmethod
        def load_train_data_from_csv(path): raise NotImplementedError("data_loader non importé")


# --- Configuration Générale ---
warnings.filterwarnings('ignore')
sns.set(style='whitegrid')
# Utiliser la seed DEPUIS config.py
utils.set_seed(config.SEED)

# --- Définir les chemins relatifs au notebook ---
# Le notebook est dans notebooks/
NOTEBOOK_DIR = Path('.') # Le répertoire courant du notebook
DATA_DIR_REL = NOTEBOOK_DIR / '../data'
PROCESSED_DATA_DIR_REL = DATA_DIR_REL / 'processed'
MODELS_SAVE_DIR_REL = NOTEBOOK_DIR / '../models_trained'

# Chemin spécifique du fichier CSV d'entraînement
TRAIN_CSV_PATH = PROCESSED_DATA_DIR_REL / 'train_processed.csv'

# Créer le dossier pour sauvegarder les modèles s'il n'existe pas
MODELS_SAVE_DIR_REL.mkdir(parents=True, exist_ok=True)

print("\n--- Configuration ---")
print(f"TensorFlow Version: {tf.__version__}")
print(f"Seed utilisée (depuis config.py): {config.SEED}")
print(f"Chemin données entraînement: {TRAIN_CSV_PATH}")
print(f"Dossier sauvegarde modèles: {MODELS_SAVE_DIR_REL}")
print(f"Dimensions Image (depuis config.py): {config.IMG_ROWS}x{config.IMG_COLS}")
```

```

# Charger les données d'entraînement en utilisant le chemin défini ci-dessus
# La fonction load_train_data_from_csv vient de src/data_loader.py
print(f"Tentative de chargement depuis : {TRAIN_CSV_PATH}")
X_train_full, y_train_full, train_ids_full = data_loader.load_train_data_from_csv(TRAIN_CSV_PATH)

if X_train_full is None:
    raise ValueError(f"Le chargement des données d'entraînement a échoué. Vérifiez le chemin '{TRAIN_CSV_PATH}' et le fichier CSV.")

# Vérifier les dimensions
print(f"Shape de X_train_full: {X_train_full.shape}")
print(f"Shape de y_train_full: {y_train_full.shape}")
print(f"Nombre d'échantillons: {X_train_full.shape[0]}")
print(f"Nombre de features: {X_train_full.shape[1]}")

# Vérifier si le nombre de features correspond aux dimensions de l'image attendue (issues de config.py)
expected_features = config.IMG_ROWS * config.IMG_COLS
if X_train_full.shape[1] != expected_features:
    raise ValueError(f"Le nombre de features ({X_train_full.shape[1]}) ne correspond pas "
                     f"aux dimensions attendues de l'image ({config.IMG_ROWS}x{config.IMG_COLS}={expected_features}). "
                     f"Vérifiez config.py ou le fichier {TRAIN_CSV_PATH}.")

# Afficher la distribution des classes
print("\nDistribution des classes :")
class_counts = pd.Series(y_train_full).value_counts(normalize=True)
print(class_counts)
plt.figure(figsize=(6, 4))
sns.countplot(x=y_train_full)
plt.title('Distribution des classes dans le jeu d'entraînement complet')
plt.xlabel('Classe')
plt.ylabel('Nombre d'échantillons')
plt.show()

num_classes = len(np.unique(y_train_full))
print(f"\nNombre de classes détectées : {num_classes}")
```


## 3. Prétraitement des Données

1.  **Séparation Train/Validation :** Nous séparons les données en ensembles d'entraînement et de validation en utilisant une stratification pour conserver la proportion des classes.
2.  **Mise à l'échelle (Scaling) :** Nous utilisons `StandardScaler` pour centrer et réduire les features. Le scaler est ajusté (`fit`) **uniquement** sur l'ensemble d'entraînement pour éviter toute fuite de données de l'ensemble de validation.


```

# Séparer en ensembles d'entraînement et de validation (ex: 80/20)
# Stratify=y_train_full est crucial à cause du déséquilibre
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full,
    test_size=0.2,         # 20% pour la validation
    random_state=config.SEED, # Utiliser SEED de config.py
    stratify=y_train_full  # Important pour le déséquilibre
)

print(f"Taille de l'ensemble d'entraînement: {X_train.shape[0]} échantillons")
print(f"Taille de l'ensemble de validation: {X_val.shape[0]} échantillons")

# Mise à l'échelle des features
scaler = StandardScaler()

# Ajuster sur X_train et transformer X_train et X_val
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val) # Utiliser le même scaler ajusté sur X_train

print("\nDimensions après mise à l'échelle:")
print(f"  X_train_scaled: {X_train_scaled.shape}")
print(f"  X_val_scaled: {X_val_scaled.shape}")

# Sauvegarder le scaler pour une utilisation future (ex: sur les données test)
# Le chemin est relatif au notebook -> ../models_trained/cnn_scaler.joblib
# scaler_save_path = MODELS_SAVE_DIR_REL / 'cnn_scaler.joblib'
# utils.save_object(scaler, scaler_save_path) # Décommentez pour sauvegarder en utilisant utils.py
```


### 3.1 Remodelage (Reshape) pour le CNN

Les CNN attendent des données d'entrée avec une structure spatiale (comme des images). Nous devons remodeler notre vecteur de features 1D (longueur 3456) en une matrice 2D (selon `config.IMG_ROWS` x `config.IMG_COLS`). Nous ajoutons également une dimension pour le canal (1 pour les données non RGB).

Le format attendu par TensorFlow/Keras est `(nombre_échantillons, hauteur, largeur, canaux)`.


```

# Remodeler les données pour le format CNN (samples, height, width, channels)
# Utiliser les dimensions IMG_ROWS et IMG_COLS depuis config.py
X_train_reshaped = X_train_scaled.reshape((-1, config.IMG_ROWS, config.IMG_COLS, 1))
X_val_reshaped = X_val_scaled.reshape((-1, config.IMG_ROWS, config.IMG_COLS, 1))

print("Dimensions après remodelage :")
print(f"  X_train_reshaped: {X_train_reshaped.shape}")
print(f"  X_val_reshaped: {X_val_reshaped.shape}")

# Visualiser un exemple d'échantillon remodelé
plt.figure(figsize=(5, 4))
plt.imshow(X_train_reshaped[0, :, :, 0], cmap='viridis') # Afficher le premier échantillon
plt.title(f'Exemple d'échantillon après reshape ({config.IMG_ROWS}x{config.IMG_COLS}) - Classe {y_train[0]}')
plt.xlabel("Feature Col Index")
plt.ylabel("Feature Row Index")
plt.colorbar(label='Valeur Scalée')
plt.show()
```


## 4. Définition du Modèle CNN

Nous construisons un modèle CNN séquentiel basé sur les paramètres définis dans `src/config.py` (`CNN_FILTERS`, `CNN_KERNEL_SIZE`, etc.). L'architecture typique comprend :
- Couches de Convolution 2D (`Conv2D`) avec activation ReLU pour extraire les features spatiales.
- Couches de Max Pooling (`MaxPooling2D`) pour réduire la dimensionnalité.
- Couches de Batch Normalization après les convolutions pour stabiliser et accélérer l'entraînement.
- Dropout pour la régularisation.
- Une couche `Flatten`.
- Des couches Denses (`Dense`) pour la classification finale.
- La couche de sortie utilise une activation `softmax`.

La fonction de perte est `sparse_categorical_crossentropy` (`config.CNN_LOSS`) car nos étiquettes `y` sont des entiers.


```

def build_cnn_model(input_shape, num_classes):
    model = keras.Sequential(name="CNN_Model")

    # Input Layer
    model.add(layers.Input(shape=input_shape))

    # Utiliser les hyperparamètres depuis config.py
    # Convolutional Blocks (Conv -> BatchNormalization -> Activation -> Pool -> Dropout)
    for i, filters in enumerate(config.CNN_FILTERS):
        model.add(layers.Conv2D(filters=filters,
                                kernel_size=config.CNN_KERNEL_SIZE,
                                padding='same', # 'same' pour conserver les dimensions après conv
                                name=f'conv2d_{i+1}'))
        model.add(layers.BatchNormalization(name=f'batchnorm_{i+1}'))
        model.add(layers.Activation(config.CNN_ACTIVATION, name=f'activation_{i+1}'))
        model.add(layers.MaxPooling2D(pool_size=config.CNN_POOL_SIZE, name=f'maxpool_{i+1}'))
        # Optionnel : Dropout après pooling
        # model.add(layers.Dropout(config.CNN_DROPOUT_RATE / 2, name=f'dropout_conv_{i+1}'))

    # Flatten Layer
    model.add(layers.Flatten(name='flatten'))

    # Dense Blocks (Dense -> Activation -> Dropout)
    for i, units in enumerate(config.CNN_DENSE_UNITS):
        model.add(layers.Dense(units, name=f'dense_{i+1}'))
        model.add(layers.Activation(config.CNN_ACTIVATION, name=f'activation_dense_{i+1}'))
        model.add(layers.Dropout(config.CNN_DROPOUT_RATE, name=f'dropout_dense_{i+1}'))

    # Output Layer
    model.add(layers.Dense(num_classes, activation=config.CNN_OUTPUT_ACTIVATION, name='output'))

    # Compiler le modèle avec les paramètres de config.py
    optimizer = config.CNN_OPTIMIZER
    loss = config.CNN_LOSS
    metrics = ['sparse_categorical_accuracy'] # Utiliser 'accuracy' ou 'sparse_categorical_accuracy'

    model.compile(optimizer=optimizer,
                  loss=loss,
                  metrics=metrics)

    return model

# Construire le modèle
input_shape = (config.IMG_ROWS, config.IMG_COLS, 1)
model = build_cnn_model(input_shape, num_classes)

# Afficher le résumé du modèle
model.summary()

# Optionnel : Visualiser l'architecture
# architecture_png_path = MODELS_SAVE_DIR_REL / 'cnn_model_architecture.png'
# tf.keras.utils.plot_model(model, show_shapes=True, to_file=architecture_png_path)
```


## 5. Gestion du Déséquilibre des Classes

Le dataset est déséquilibré. Nous utilisons la pondération des classes (`class_weight`) pendant l'entraînement pour donner plus d'importance aux classes minoritaires. `sklearn.utils.class_weight.compute_class_weight` calcule les poids.


```

# Calculer les poids des classes
class_weights = class_weight.compute_class_weight(
    'balanced',
    classes=np.unique(y_train),
    y=y_train
)

# Créer un dictionnaire pour Keras
class_weights_dict = dict(enumerate(class_weights))

print("Poids calculés pour chaque classe :")
for cls, weight in class_weights_dict.items():
    print(f"  Classe {cls}: {weight:.4f}")
```


## 6. Entraînement du Modèle

Nous entraînons le modèle.
- **Callbacks:**
    - `EarlyStopping`: Arrête l'entraînement si `val_loss` ne s'améliore plus.
    - `ModelCheckpoint`: Sauvegarde le meilleur modèle (basé sur `val_loss`) pendant l'entraînement dans le dossier `models_trained/` (chemin relatif: `../models_trained/`).


```

# Définir les callbacks
# Early Stopping
early_stopping = callbacks.EarlyStopping(
    monitor='val_loss',      # Métrique à surveiller
    patience=15,             # Nb d'epochs sans amélioration avant d'arrêter
    restore_best_weights=True, # Revenir aux poids du meilleur epoch
    verbose=1
)

# Model Checkpoint
# Utiliser le chemin défini plus haut (relatif au notebook)
model_checkpoint_path = MODELS_SAVE_DIR_REL / 'best_cnn_model.keras'
model_checkpoint = callbacks.ModelCheckpoint(
    filepath=model_checkpoint_path,
    monitor='val_loss',
    save_best_only=True, # Ne sauvegarder que le meilleur modèle
    verbose=1
)

# Paramètres d'entraînement
EPOCHS = 100 # Nombre maximal d'epochs (EarlyStopping peut arrêter avant)
BATCH_SIZE = 32

print("Début de l'entraînement...")
print(f"Le meilleur modèle sera sauvegardé dans: {model_checkpoint_path}")

# Entraîner le modèle
history = model.fit(
    X_train_reshaped, y_train,
    validation_data=(X_val_reshaped, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weights_dict,
    callbacks=[early_stopping, model_checkpoint], # Utiliser les callbacks définis
    verbose=1 # Mettre à 2 pour une sortie par batch, 1 par epoch, 0 silencieux
)

print("\nEntraînement terminé.")

# Le meilleur modèle est automatiquement restauré par EarlyStopping(restore_best_weights=True)
# Ou peut être chargé manuellement depuis le fichier .keras sauvegardé par ModelCheckpoint si besoin :
# print(f"Chargement du meilleur modèle depuis: {model_checkpoint_path}")
# model = keras.models.load_model(model_checkpoint_path)
```


## 7. Évaluation du Modèle

Nous évaluons les performances du meilleur modèle sur l'ensemble de validation.

1.  **Courbes d'Apprentissage :** Loss et Accuracy (via `utils.plot_history`).
2.  **Évaluation Finale :** Loss et Accuracy sur la validation.
3.  **Rapport de Classification.**
4.  **Accuracy Équilibrée (Balanced Accuracy).**
5.  **Matrice de Confusion** (via `utils.plot_confusion_matrix`).


```

# 1. Courbes d'Apprentissage (utilise la fonction de utils.py)
print("\nAffichage des courbes d'apprentissage:")
utils.plot_history(history) # Utilise la fonction importée depuis src/utils.py

# 2. Évaluation Finale sur l'ensemble de validation
print("\nÉvaluation sur l'ensemble de validation:")
loss, accuracy = model.evaluate(X_val_reshaped, y_val, verbose=0)
print(f"  Loss: {loss:.4f}")
print(f"  Accuracy: {accuracy:.4f}")

# 3. Prédictions et Rapport de Classification
y_pred_proba = model.predict(X_val_reshaped)
y_pred = np.argmax(y_pred_proba, axis=1) # Obtenir la classe prédite

print("\nRapport de Classification :")
print(classification_report(y_val, y_pred, target_names=[f'Classe {i}' for i in np.unique(y_train_full)]))

# 4. Balanced Accuracy
balanced_acc = balanced_accuracy_score(y_val, y_pred)
print(f"Accuracy Équilibrée (Balanced Accuracy): {balanced_acc:.4f}")

# 5. Matrice de Confusion (utilise la fonction de utils.py)
print("\nMatrice de Confusion :")
utils.plot_confusion_matrix(y_val, y_pred, classes=np.unique(y_train_full)) # Utilise la fonction importée
```
